```python
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon
import numpy as np

# Hàm vẽ ngôi sao
def draw_star(ax, center_x, center_y, size, color='yellow'):
    # Tính toán vị trí các điểm của ngôi sao
    outer_radius = size
    inner_radius = size * 0.382  # Tỉ lệ khoảng 0.382 theo hình học ngôi sao
    angles = np.linspace(0, 2 * np.pi, 11)  # 10 điểm, 5 điểm ngoài và 5 điểm trong
    star_points = []
    for i, angle in enumerate(angles):
        radius = outer_radius if i % 2 == 0 else inner_radius
        x = center_x + radius * np.sin(angle)
        y = center_y + radius * np.cos(angle)
        star_points.append([x, y])

    star = Polygon(star_points, closed=True, color=color)
    ax.add_patch(star)

# Thiết lập nền cờ Việt Nam
def draw_vietnam_flag():
    fig, ax = plt.subplots()
    ax.set_xlim(0, 3)
    ax.set_ylim(0, 2)

    # Vẽ nền đỏ (tỉ lệ 3:2)
    ax.add_patch(plt.Rectangle((0, 0), 3, 2, color='red'))

    # Vẽ ngôi sao vàng ở giữa
    draw_star(ax, center_x=1.5, center_y=1, size=0.5)

    # Tắt trục tọa độ
    ax.set_axis_off()

    plt.gca().set_aspect('equal', adjustable='box')
    plt.show()

# Gọi hàm vẽ cờ
draw_vietnam_flag()
```
